```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

url = "https://raw.githubusercontent.com/fjmv89/Trabajo-Practico-Grupal/main/Dataset_farmacia.csv"
df = pd.read_csv(url, sep=";", encoding="cp1252", dtype=str)
df.head()
```

Para darle un análisis extra al trabajo con material dado en clases, se estudia el comportamiento de los clientes frente a los aumentos de precios mediante el calculo de la elasticidad–precio de la demanda. Para lograrlo con las variables que presenta el dataset, el equivalente práctico y metodológico adecuado utilizado fue la fórmula del punto medio o elasticidad arco.
Aunque el código esté ejecutado con soporte de la IA, las conclusiones son exclusivas del criterio humano.

```python
df_f = df.copy()
df_f['CANTID'] = pd.to_numeric(
    df_f['CANTID'].str.replace(',', '.'), errors='coerce'
)
df_f['PUBLICO VENTA'] = pd.to_numeric(
    df_f['PUBLICO VENTA'].str.replace(',', '.'), errors='coerce'
)
df_f['FECHA'] = pd.to_datetime(
    df_f['FECHA'], format='%d/%m/%Y', errors='coerce'
)

df_f = df_f.dropna(subset=['DESCRIPX', 'FECHA', 'CANTID', 'PUBLICO VENTA'])

palabras_excluidas = [
    'PAÑALES','TOALLITAS', 'JABON','TALCO', 'CREMA','COSMETICA', 'PROTECTOR', 'AGUJAS', 'JERINGAS','PRESERV','PASTA','COLONIA','PERFUME','TINTURA', 'EMULSION', "CREMA HIDRATANTE", "SERUM FACIAL", "PROTECTOR SOLAR", "AGUA MICELAR",
    "TONICO FACIAL", "GEL DE LIMPIEZA FACIAL", "BALSAMO LABIAL", "MASCARILLA FACIAL","EMULSION HIDRATANTE", "CREMA ANTIARRUGAS", "EXFOLIANTE FACIAL", "LECHE DE LIMPIEZA",
    "CREMA CORPORAL", "GEL DE DUCHA", "JABON LIQUIDO", "JABON DE TOCADOR", "ACEITE CORPORAL", "CREMA DE MANOS", "CREMA DE PIES", "TALCO PARA PIES","DESODORANTE", "ANTITRANSPIRANTE", "CREMA REDUCTORA",
    "SHAMPOO", "ACONDICIONADOR", "MASCARA CAPILAR", "TINTURA PARA EL CABELLO", "AMPOLLA CAPILAR", "SERUM CAPILAR", "CREMA PARA PEINAR", "GEL CAPILAR",  "ESPUMA DE AFEITAR", "MAQUINITA DE AFEITAR", "BANDAS DE CERA DEPILATORIA",
    "CREMA DEPILATORIA", "GEL PARA DESPUES DE AFEITAR","PASTA DE DIENTES", "CEPILLO DE DIENTES", "ENJUAGUE BUCAL", "HILO DENTAL"
]


def es_medicamento(texto):
  if not isinstance(texto, str):
    return False
  t = texto.upper()
  for ex in palabras_excluidas:
    if re.search(r'\b' + re.escape(ex) + r'\b', t):
      return False
  return True


df_f = df_f[df_f['DESCRIPX'].apply(es_medicamento)]

def extraer_nombre_base(texto):
  t = texto.upper()
  t = re.sub(r'\b\d+[\.,]?\d*\s*(MG|G|GR|ML|CC|UI|%)\b', '', t)
  t = re.sub(
      r'\b(COMP|CÁPS|CAPS|TAB|SUSP|SOL|GEL|JARABE|GOTAS|AMP|OVULOS|SPRAY|INHAL|F|FORTE|REC|MAST|LIB|PROL)\b',
      '',
      t,
  )
  t = re.sub(r'\bx\s*\d+\b', '', t)
  t = re.sub(r'[\.\,\-\/\(\)]', ' ', t)
  t = re.sub(r'\s+', ' ', t).strip()

  palabras = t.split()
  if len(palabras) >= 2:
    return f'{palabras[0]} {palabras[1]}'
  elif len(palabras) == 1:
    return palabras[0]
  return ''


df_f['DESCRIP_BASE'] = df_f['DESCRIPX'].apply(extraer_nombre_base)
df_f = df_f[df_f['DESCRIP_BASE'] != '']

df_f['Mes'] = df_f['FECHA'].dt.to_period('M')

mensual = (
    df_f.groupby(['DESCRIP_BASE', 'Mes'])
    .agg({'CANTID': 'sum', 'PUBLICO VENTA': 'mean'})
    .reset_index()
    .sort_values(['DESCRIP_BASE', 'Mes'])
)

inicio_mes = pd.Period('2025-09', freq='M')
final_mes = pd.Period('2026-09', freq='M')

df_inicio = mensual[mensual['Mes'] == inicio_mes].rename(
    columns={
        'CANTID': 'Q_inicio',
        'PUBLICO VENTA': 'P_inicio',
        'Mes': 'Mes_inicio',
    }
)
df_final = mensual[mensual['Mes'] == final_mes].rename(
    columns={'CANTID': 'Q_final', 'PUBLICO VENTA': 'P_final', 'Mes': 'Mes_final'}
)

resultado = pd.merge(df_inicio, df_final, on='DESCRIP_BASE')
resultado = resultado.dropna(
    subset=['Q_inicio', 'Q_final', 'P_inicio', 'P_final']
)

Q1 = resultado['Q_inicio']
Q2 = resultado['Q_final']
P1 = resultado['P_inicio']
P2 = resultado['P_final']

delta_q = Q2 - Q1
promedio_q = (Q1 + Q2) / 2.0

delta_p = P2 - P1
promedio_p = (P1 + P2) / 2.0

resultado['elasticidad_arco'] = np.where(
    (promedio_q != 0) & (promedio_p != 0) & (delta_p != 0),
    abs((delta_q / promedio_q) / (delta_p / promedio_p)),
    np.nan,
)

resultado.loc[(delta_q == 0) & (delta_p != 0), 'elasticidad_arco'] = 0.0
resultado = resultado.replace([np.inf, -np.inf], np.nan).dropna(
    subset=['elasticidad_arco']
)

def clasificar_punto_medio(e):
  if np.isclose(e, 1.0, atol=1e-4):
    return 'Elasticidad Unitaria (|E| = 1)'
  elif e > 1.0:
    return 'Demanda Elástica (|E| > 1)'
  else:
    return 'Demanda Inelástica (0 ≤ |E| < 1)'


resultado['clasificacion_demanda'] = resultado['elasticidad_arco'].apply(
    clasificar_punto_medio
)

print(f'\n--- UNIVERSO TOTAL AUTOMATIZADO Y LIMPIO: {len(resultado)} PRODUCTOS ---')
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
print(
    resultado[[
        'DESCRIP_BASE',
        'Q_inicio',
        'P_inicio',
        'Q_final',
        'P_final',
        'elasticidad_arco',
        'clasificacion_demanda',
    ]]
    .head(20)
    .to_string(index=False)
)

conteo_categorias = resultado['clasificacion_demanda'].value_counts()

plt.figure(figsize=(9, 5))
conteo_categorias.plot(
    kind='bar', color=['#337ab7', '#f0ad4e', '#d9534f'], edgecolor='black'
)
plt.title(
    f'Clasificación de los {len(resultado)} Productos según la Fórmula del'
    ' Punto Medio (|E|)',
    fontsize=11,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Categoría Teórica', fontsize=10, fontweight='bold')
plt.ylabel('Cantidad de Productos', fontsize=10, fontweight='bold')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle=':', alpha=0.5)
plt.tight_layout()
plt.show()
```

Para procesar los registros de forma eficiente en el código, se implementó un sistema de filtrado automático por exclusión para descartar todos aquellos artículos que no corresponden a medicamentos, como lo son los de perfumería y cosmética, evitando el uso de listas manuales restrictivas.
El recuento final arroja la cifra de 592 productos analizados en lugar de 298 estimados inicialmente, debido a que el sistema brinda la totalidad del archivo de datos disponible de forma masiva. Al no aplicar filtros manuales, se incorporan marcas secundarias y genéricas que tienen presencia comercial real, unificando sus diferentes presentaciones para medir la demanda con mayor presición.


**Procedimiento del código:** Primero se hizo una limpieza de tipos de datos, la cantidad y el precio al público se vuelven un dato numérico, la fecha se pasa al formato estándar y se eliminaron valores faltantes. Luego se hizo el filtro de exclusión y la limpieza automática de la columna "DESCRIPX", se creó el período mensual para agrupar los productos vendidos en cada uno. Se realizó el inicio de mes (septiembre de 2025) y el último (septiembre de 2026) para aplizar la fórmula del punto medio; lo mismo con el precio. Se definió las condiciones para cada caso y se finaliza con el gráfico de barras

**Conclusiones:** La segmentación de la demanda según las categorías permite identificar qué cantidad de productos muestran sensibilidad ante los incrementos en el precio (demanda elástica), cuáles mantienen su consumo a pesar de que suba el precio (demanda inelástica; esto es más normal en un rubro de la salud, son bienes necesarios si se tratan de medicamentos para enfermedades graves) o si se ubican en el umbral de la elasticidad unitaria.

**Gestión:** Comparado con el IPC calculado anteriormente, la gerencia cuenta con margen para ajustar selectivamente aquellos precios que quedaron retrasados, siempre evaluando la sensibilidad del consumidor obtenida en la elasticidad (esto porque el índice de precios se mantuvo por debajo de la inflación acumulada general).
Para los productos clasificados con demanda inelástica, la empresa puede trasladar los incrementos necesarios alineados con el IPC sin correr el riesgo re perder volumen de ventas de manera drástica. Por el contrario, con loa productos con una demanda elástica, cualquier intento de subir o equiparar de golpe la inflación general generaria una contradicción en las cantidades vendidas; por lo tanto, se deben priorizar acuerdos, promociones o aumentos más moderados que el IPC.

**Limitaciones:** El análisis realizado se limita exclusivamente a la información contenida en este único conjunto de datos, por lo que los resultados no deben interpretarse como una representación absoluta de todo el mercado. No es posible deducir conclusiones definitivas ni generalizaciones a nivel microecónomico sobre este rubro.
